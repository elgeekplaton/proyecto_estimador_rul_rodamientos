```python
#Verificamos contenido de la carpeta donde están los archivos del modelo y dataset.
%cd /content/sample_data/
!ls
```

```python
#Iniciamos streamlit
import subprocess
import time

with open('/content/sample_data/streamlit_log.txt', 'w') as f:
    proceso = subprocess.Popen(
        ['streamlit', 'run', 'app.py', '--server.port', '8501', '--server.headless', 'true'],
        stdout=f, stderr=subprocess.STDOUT,
        cwd='/content/sample_data/'
    )
time.sleep(8)
print('Streamlit vivo:', proceso.poll())   # None = sigue corriendo, sin bloquear la celda
```

```python
#Verificación de servicio activo de streamlit (Opcional - None significa Ok)
print('¿Streamlit sigue corriendo?', proceso.poll())
```

```python
#Verificando si del lado del servidor esta activo el acceso web al modelo (OK si retorna 200)
!curl -s -o /dev/null -w "%{http_code}" http://localhost:8501
```
